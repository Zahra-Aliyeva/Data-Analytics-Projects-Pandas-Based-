```python
import pandas as pd
heart_disease=pd.read_csv("heart-disease.csv")
heart_disease.head()
```

```python
heart_disease.target.value_counts()
```

```python
heart_disease['target'].value_counts().plot(kind='bar')
```

```python
heart_disease.hist(figsize=(10,10))
```

```python
heart_disease['age'].hist(bins=20)
```

```python
heart_disease.plot.scatter(x='age',y='chol')
```

```python
import matplotlib.pyplot as plt
colors=['red' if t==1 else 'blue' for t in heart_disease ['target']]

plt.scatter(
    heart_disease['age'],
    heart_disease['chol'],
    c=colors
);
```

```python
fig, ax = plt.subplots(figsize=(10,6))
scatter = ax.scatter(
x = heart_disease['age'],
y = heart_disease['chol'],
c = heart_disease['target']
)
plt.colorbar(scatter)
ax.set(title='Heart disease and cholesterol levels',
xlabel='Age',
ylabel='Cholesterol');
```

```python
plt.figure(figsize=(6,5))
heart_disease.boxplot(column='age', by='target')
plt.title('Age vs Heart Disease')
plt.suptitle('')
plt.xlabel('Target (0=No, 1=Yes)')
plt.ylabel('Age')
plt.show();
```

```python
corr= heart_disease.corr()
corr
```

```python
corr = heart_disease.corr()
plt.figure(figsize=(10,8))
plt.imshow(corr)
plt.colorbar()
plt.title('Correlation Matrix')
plt.xticks (range(len(corr)), corr.columns, rotation=90)
plt.yticks (range (len(corr)), corr.columns)
plt.show()
```

This dataset covers 303 patients and it's fairly balanced, 165 with disease and 138 without, so the results aren't skewed toward one side. At first glance, simple indicators like age and cholesterol don't show any clear relationship, even coloring them by disease status doesn't reveal a pattern. One interesting thing is that the median age for patients with disease (52) is actually lower than those without (58), which is the opposite of what you'd expect. The real signal shows up in the correlation analysis, where the strongest links to disease come from things measured during exercise or a stress test, like cp (chest pain type), exang (exercise-induced angina), oldpeak, and thalach. Cholesterol, barely correlates with anything, and age is weak too. So in this dataset, the "classic" risk factors don't carry as much weight as expected, the real differentiators are the functional/exercise-based measurements.
